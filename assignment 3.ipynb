# Q11

```python
even_sum = 0
odd_sum = 0

for i in range(1, 51):
    if i % 2 == 0:
        even_sum = even_sum + i
    else:
        odd_sum = odd_sum + i

print("Sum of even numbers:", even_sum)
print("Sum of odd numbers:", odd_sum)
```

## WHILE LOOP

```python
even_sum = 0
odd_sum = 0
i = 1

while i <= 50:
    if i % 2 == 0:
        even_sum = even_sum + i
    else:
        odd_sum = odd_sum + i

    i = i + 1

print("Sum of even numbers:", even_sum)
print("Sum of odd numbers:", odd_sum)
```

# Q14

```python
number_of_guests = 300

discard_menu = ["fish", "qeema", "nihari"]

menu = []
menu_price = {}
menu_price_per_head = {}

for i in range(5):
    dish = input("Enter dish name: ")

    if dish.lower() in discard_menu:
        print("This dish is not allowed.")
        continue

    price = float(input("Enter total price for 300 guests: "))

    menu.append(dish)
    menu_price[dish] = price

total_menu_price = sum(menu_price.values())

for dish in menu:
    menu_price_per_head[dish] = menu_price[dish] / number_of_guests

total_per_head_cost = sum(menu_price_per_head.values())

print("Menu:", menu)
print("Menu Prices:", menu_price)
print("Total Menu Price:", total_menu_price)
print("Per Head Prices:", menu_price_per_head)
print("Complete Menu Per Head Cost:", total_per_head_cost)
```

## WHILE LOOP

```python
number_of_guests = 300

discard_menu = ["fish", "qeema", "nihari"]

menu = []
menu_price = {}
menu_price_per_head = {}

i = 0

while i < 5:
    dish = input("Enter dish name: ")

    if dish.lower() in discard_menu:
        print("This dish is not allowed.")
        i = i + 1
        continue

    price = float(input("Enter total price for 300 guests: "))

    menu.append(dish)
    menu_price[dish] = price

    i = i + 1

total_menu_price = sum(menu_price.values())

for dish in menu:
    menu_price_per_head[dish] = menu_price[dish] / number_of_guests

total_per_head_cost = sum(menu_price_per_head.values())

print("Menu:", menu)
print("Menu Prices:", menu_price)
print("Total Menu Price:", total_menu_price)
print("Per Head Prices:", menu_price_per_head)
print("Complete Menu Per Head Cost:", total_per_head_cost)
```
